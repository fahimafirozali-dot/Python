```python
#program to get the 3rd element and 3rd element from last of a tuple
numbers=(3,5,7,2,5,1,9)
print("The third number:",numbers[2])
print("The third from last number",numbers[-3])
```

```python
#program to accept sequence of comma seperated numbers from user and generate a tuple with those numbers and find average of all items in that tuple
n=input("Enter the numbers:")
tuple = tuple(int(x) for x in n.split(","))
average = sum(tuple) / len(tuple)
print("Tuple:",tuple)
print("Average:",average)
```
