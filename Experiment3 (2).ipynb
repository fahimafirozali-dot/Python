```python
#program to change a given string to new string where first and last char have been exchanged
string=input("Enter a string:")
new_string=string[-1]+string[1:-1]+string[0]
print("New string is:",new_string)
```

```python
#program to get string made of first two and last two chars from a given string
string=input("Enter a string:")
new_string=string[:2]+string[-2:]
print("new string:",new_string)
```

```python
#program to create a single string seperated with space from two strings by swapping char at pos 1
string1=input("Enter a string:")
string2=input("Enter a string:")
new_string1=string2[0]+string1[1:]
new_string2=string1[0]+string2[1:]
print(new_string1,new_string2)
```

```python
#program to get string from given string where all occurences of its first char have been changed to $ except first char itself
string=input("Enter a string:")
new_string=string[0]+string[1:].replace(string[0],"$")
print("new string is:",new_string)
```
