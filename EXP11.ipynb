```python
#Generate positive list of numbers from a given list of integers 
numbers=list(map(int,input("Enter the list:").split()))
print("The list:",numbers)
result=[i for i in numbers if i>0]
print("Positive numbers are",result)
```

```python
#Square of N numbers 
n=int(input("Enter N:"))
print("N is:",n)
result=[i*i for i in range(1,n+1)]
print("Squares:",result)
```

```python
#Form a list of vowels selected from a given word 
word=input("Enter a word:")
print("The word is :",word)
results = [i for i in word if i.lower() in "aeiou"]
print("Vowels:",results)
```

```python
#List ordinal value of each element of a word (Hint: use ord() to get ordinal values)
word=input("Enter the word:")
print("The word is:",word)
result=[ord(i) for i in word]
print("Ordinal  values:",result)
```
