```python
#program to accept 5 students name and roll number,and map these into dictionary.Sort the dictionary in asc and desc order of their roll no
roll_no=input("Enter 5 roll no:").split()
name=input("Enter 5 name:").split()
students=dict(zip(roll_no,name))
ascending=dict(sorted(students.items()))
descending=dict(sorted(students.items(),reverse=True))
print("dictionary",students)
print("ascending order:",ascending)
print("descending order:",descending)
```

```python
#program to accept 2 dictionary.Merge two dictionaries and display it
dict1={}
dict2={}
roll1=input("Enter 3 roll numbers:").split()
name1=input("Enter 3 students name:").split()
for r,n in zip(roll1,name1):
    dict1[r]=n
roll2=input("Enter 2 roll numbers:").split()
name2=input("Enter 2 students name:").split()
for r,n in zip(roll2,name2):
    dict2[r]=n
    dict1.update(dict2)
    print("Merged dictionary:",dict1)
```
