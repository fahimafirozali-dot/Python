```python
#program to create a list of colors from comma-seperated color names entered by user.Display first and last colors.
color=input("Enter color names:")
c=color.split(",")
print("First color:",c[0])
print("Last color:",c[-1])
```

```python
#program to accept a filename from the user,print extesion of that.
file_name=input("Enter the file name:")
extension=file_name.split(".")[-1]
print("Extension",extension)
```

```python
#program which accepts a sequence of comma seperated nos from users and generate a list with those nos and find sum of all items in list.
numbers=input("Enter the list of numbers:")
list=[int(x)for x in numbers.split(",")]
print("The sum is:",sum(list))
```
