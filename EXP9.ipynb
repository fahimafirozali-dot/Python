```python
#program to add "ing" to the end of a given string. If the string already ends with "ing", add "ly" instead.
s=input("Enter a string:")
if s[-3:]=="ing":
    print(s+"ly")
else:
    print(s+"ing")
```

```python
#program to check the given string is palindrome or not.
s=input("Enter a string:")
if s==s[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```
