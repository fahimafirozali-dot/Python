```python
#program to print all colors from colorset1 not contained in colorset2
colorset1=input("Enter colors of colorset1")
colorset2=input("Enter colors of colorset2")
c1=set(colorset1.split(","))
c2=set(colorset2.split(","))
c=c1-c2
print("Colorset1 :",c1)
print("Colorset2 :",c2)
print("Output is:",c)
```

```python
#program to print all colors from colorset1 and colorset2
colorset1=input("Enter colors of colorset1")
colorset2=input("Enter colors of colorset1")
c1=set(colorset1.split(","))
c2=set(colorset2.split(","))
c=c1 | c2
print("Colorset1 :",c1)
print("Colorset2 :",c2)
print("Output is:",c)
```
