```python
import os 
os.makedirs("graphics/graphics3D",exist_ok=True)
open("graphics/__init__.py","w").close()
open("graphics/graphics3D/__init__.py","w").close()
```

```python
%%writefile graphics/rectangle.py
#Rectangle module.
def area(l,b):
    return l*b
def perimeter(l,b):
    return 2*(l+b)
```

```python
%%writefile graphics/circle.py
#Circle module.
import math
def area(r):
    return math.pi*r*r
def perimeter(r):
    return 2*math.pi*r
```

```python
%%writefile graphics/graphics3D/cuboid.py
#Cube module.
def area(l,b,h):
    return 2*(l*b+l*h+b*h)
def perimeter(l,b,h):
    return 4*(l+b+h)
```

```python
%%writefile graphics/graphics3D/sphere.py
#Sphere module.
import math
def area(r):
    return 4*math.pi*r*r
def perimeter(r):
    return 2*math.pi*r
```

```python
# Importing modules
import graphics.rectangle
import graphics.circle
from graphics.graphics3D import cuboid, sphere

while True:
    print("\n----- GRAPHICS MENU -----")
    print("1. Rectangle")
    print("2. Circle")
    print("3. Cuboid")
    print("4. Sphere")
    print("5. Exit")

    ch = int(input("Enter your choice: "))

    if ch == 1:
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))

        print("Area of Rectangle:", graphics.rectangle.area(l, b))
        print("Perimeter of Rectangle:", graphics.rectangle.perimeter(l, b))

    elif ch == 2:
        r = float(input("Enter radius: "))

        print("Area of Circle:", graphics.circle.area(r))
        print("Circumference of Circle:", graphics.circle.perimeter(r))

    elif ch == 3:
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))
        h = float(input("Enter height: "))

        print("Total Surface Area of Cuboid:", cuboid.area(l, b, h))
        print("Perimeter of Cuboid:", cuboid.perimeter(l, b, h))

    elif ch == 4:
        r = float(input("Enter radius: "))

        print("Surface Area of Sphere:", sphere.area(r))
        print("Circumference of Sphere:", sphere.perimeter(r))

    elif ch == 5:
        print("Exiting the program...")
        break

    else:
        print("Invalid choice! Please try again.")
```
