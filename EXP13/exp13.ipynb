```python
#create two modules and import them
import Rectangle
import Circle

l = float(input("Enter the length:"))
b = float(input("Enter the breadth:"))
r = float(input("Enter the radius:"))

print("Length:", l)
print("Breadth:", b)
print("Radius:", r)

print(f"Area of circle:{Circle.area(r):.2f}")
print(f"Perimeter of circle:{Circle.perimeter(r):.2f}")

print(f"Area of Rectangle:{Rectangle.area(l,b):.2f}")
print(f"Perimeter of Rectangle:{Rectangle.perimeter(l,b):.2f}")
```
