```python
#program to print all colors from colorset1 not contained in colorset2
colorset1={"red","yellow","blue","green","purple"}
colorset2={"black","red","white","blue","green"}
new_set=(colorset1-colorset2)
print("Output is:",new_set)
```

```python
#program to print all colors from colorset1 and colorset2
colorset1={"red","yellow","blue","green","purple"}
colorset2={"black","red","white","blue","green"}
new_set=colorset1 | colorset2
print("Output is:",new_set)
```
