```python
#occurence of each word in aline of text

s=input("Enter a line: ")
words=s.split()

for word in set(words):
    print(word,":", words.count(word))
```

```python
#in a list of integers, for all value greater than 100 store 'over' instead.

a=list(map(int, input("Enter numbers: ").split()))

for i in range(len(a)):
    if a[i]>100:
        a[i] = "over"
print(a)
```

```python
#count the occurences of a within the list
names = ["Anamika","Athira","Amrutha","Raju"]

count =0

for name in names:
    count=count +name.lower().count("a")
    
print("Occurences of a:",count)
    
```

```python
#find gcd of two numbers
a=int(input("Enter first number: "))
b=int(input("Enter second number: "))

while b:
    a,b=b,a%b
print("GCD:",a)
```

```python
#create a list removing even numbers
a=list(map(int,input("Enter numbers:").split()))
b=[]
for x in a:
    if x % 2 !=0:
        b.append(x)
print("List after removing even numbers:",b)
```

```python
#factorial of a number
n=int(input("Enter a number: "))
fact=1
for i in range(1,n+1):
    fact= fact * i
print("Factorial:", fact)
```

```python
#find the sum of list
numbers = list(map(int,input("Enter the numbers (seperated by space):").split()))
sum=0
print("The numbers are:",numbers)
for i in numbers:
    sum=sum+i
print("Sum of all items:",sum)
```

```python
#Find four-digit perfect square with all digits even
import math
start=int(input("Enter the starting range:"))
end=int(input("Enter the ending range:"))
print("Starting range:",start,"\nEnding range:",end)
print("Result is:",end=" ")
for n in range(start,end+1):
    if int(math.sqrt(n))**2==n:
        if all(d in "02468" for d in str(n)):
            print(n,end=" ")
```

```python
#count the character frequency in a string
s=input("Enter a string:")
print("String is:",s)
print("Character frequency:")
chars=[]
for ch in s:
    if ch not in chars:
        chars.append(ch)
for ch in chars:
    print(ch, ":", s.count(ch))
```

```python
#Compare two lists by length,sum and common value
list1=list(map(int,input("Enter first list:").split()))
list2=list(map(int,input("Enter second list:").split()))
print("First list:",list1)
print("Second list:", list2)
if len(list1)==len(list2):
    print("Both list have the same length.")
else:
    print("Both list do not have the same length.")
if sum(list1)==sum(list2):
    print("Both list have the same sum")
else:
    print("Both list do not have the same sum.")
if set(list1)&set(list2):
    print("Both list have a common value.")
else:
    print("Both list do not have a common value.")
```

```python
#Generate a number pyramid for n rows
n=int(input("Enter the number:"))
print("The number:",n)
for i in range(1,n+1):
    for j in range(1,i+1):
        print(i*j,end=" ")
    print()
        
```
